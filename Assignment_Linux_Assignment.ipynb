1. What is Linux?

Answer -> Linux is an open-source, Unix-like operating system developed by Linus Torvalds in 1991. It acts as an interface between the user and the computer hardware. Linux is widely used in servers, cloud computing, embedded systems, supercomputers, and desktops because it is secure, stable, and free.

2. Difference between Hard Link and Soft Link:

Hard Link
1. A hard link is another name for the same file.
2. Both the original file and the hard link point to the same inode.

Example:
1. Create a file-->  touch file1  echo "gdgad" > file1

2. Create Hard Link--> Create Hard Link

3. Check inode numbers--> ls -li

Soft Link (Symbolic Link)

1. A soft link is like a shortcut to the original file.
2. It stores the path of the target file.

Example

1. Create a file -->  touch file2   echo "gdjasgdjka" > file2

2. Create Soft Link--> ln -s file2 softlink

3. Check link--> ls -l


3. What is Kernel in Linux?

Answer

The Kernel is the core component of the Linux operating system. It manages communication between hardware and software.

Whenever a user runs a command, the request first goes to the kernel. The kernel interacts with the hardware and returns the result to the user.

Functions of Kernel
1. Process Management
2. Memory Management
3. Device Management
4. File System Management
5. CPU Scheduling
6. Security and Permissions


Diagram

4. HOW DO YOU CREATE A USER ACCOUNT IN LINUX?

A user account in Linux can be created using the useradd command. After creating the account, a password must be assigned so that the user can log in to the system.

STEP 1: CREATE A USER

sudo useradd p1

STEP 2: SET A PASSWORD

sudo passwd p1

Example Output:

Enter new password:
Retype new password:
passwd: password updated successfully

STEP 3: VERIFY THE USER ACCOUNT

id p1

Example Output:

uid=1002(p1)
gid=1002(p1)
groups=1002(p1)

CHECK ALL USERS IN THE SYSTEM

cat /etc/passwd

Example Output:

root:x:0:0:root:/root:/bin/bash
daemon:x:1:1:daemon:/usr/sbin:/usr/sbin/nologin
p1:x:1002:1002::/home/p1:/bin/sh

COMPLETE EXAMPLE

Create User:

sudo useradd p1

Set Password:

sudo passwd p1

Verify User:

id p1

Output:

uid=1002(p1)
gid=1002(p1)
groups=1002(p1)

RESULT

The user account p1 is successfully created, assigned a password, and verified in the system.

5. WHAT IS THE GREP COMMAND USED FOR IN LINUX?

The grep command is used to search for specific text or patterns inside files. It displays the lines that match the given pattern.

The word grep stands for:

GLOBAL REGULAR EXPRESSION PRINT

1. CREATE A SAMPLE FILE

cat > fruits.txt

Type the following data:

Apple
Banana
Orange
Apple Juice
Mango

Press:

Ctrl + D

2. SEARCH FOR A WORD

grep Apple fruits.txt

OUTPUT:

Apple
Apple Juice

3. IGNORE CASE WHILE SEARCHING

grep -i apple fruits.txt

OUTPUT:

Apple
Apple Juice

4. SHOW LINE NUMBERS

grep -n Apple fruits.txt

OUTPUT:

1:Apple
4:Apple Juice

5. COUNT THE NUMBER OF MATCHES

grep -c Apple fruits.txt

OUTPUT:

2

6. RECURSIVE SEARCH IN DIRECTORIES

grep -r "password" .

This command searches for the word password in all files and subdirectories of the current directory.

7. COMPLETE EXAMPLE

cat > fruits.txt
Apple

Apple Juice
Mango
Ctrl + D

grep Apple fruits.txt

grep -c Apple fruits.txt

8. RESULT

The grep command is a powerful text-searching utility in Linux that helps users find words, patterns, line numbers, and occurrence counts within files and directories.

6. CREATE USER p1 AND ADD TO GROUPS

● REQUIREMENT
Create user p1
Add user to groups g1, g2, g3
Default file group should be g1 (primary group)

● STEP 1: CREATE GROUPS
sudo groupadd g1
sudo groupadd g2
sudo groupadd g3

● STEP 2: CREATE USER WITH GROUPS
sudo useradd -g g1 -G g2,g3 p1

Explanation
-g → primary group
-G → secondary groups

Primary group: g1
Secondary groups: g2, g3

● STEP 3: SET PASSWORD
sudo passwd p1

● STEP 4: VERIFY USER
id p1

Output
uid=1005(p1)
gid=1001(g1)
groups=1001(g1),1002(g2),1003(g3)

● STEP 5: LOGIN USER
su - p1

● STEP 6: CREATE FILE AND CHECK GROUP
touch demo.txt
ls -l

Output
-rw-r--r-- 1 p1 g1 demo.txt

● RESULT
User p1 is created successfully and added to multiple groups. Since g1 is the primary group, all created files automatically belong to g1.

7. CREATE DIRECTORY PERMISSION ASSIGNMENT

● REQUIREMENT
Create directory /tmp/bg
Owner should be abhi
Owner should be able to: create files, delete files, read files, modify files

● STEP 1: CREATE DIRECTORY
sudo mkdir /tmp/bg

● STEP 2: CREATE USER (IF NOT EXISTS)
sudo useradd abhi
sudo passwd abhi

● STEP 3: CHANGE OWNER
sudo chown abhi:abhi /tmp/bg

● STEP 4: SET PERMISSIONS
sudo chmod 700 /tmp/bg

Permission Meaning
Owner → rwx (full access)
Group → --- (no access)
Others → --- (no access)

● STEP 5: LOGIN USER
su - abhi

● STEP 6: MOVE TO DIRECTORY
cd /tmp/bg

● STEP 7: CREATE FILE
touch file1

● STEP 8: WRITE INTO FILE
echo Hello > file1

● STEP 9: APPEND DATA
echo Linux >> file1

● STEP 10: VIEW FILE
cat file1

● STEP 11: DELETE FILE
rm file1

● RESULT
The directory /tmp/bg is successfully created and assigned to user abhi. Only the owner has full control (read, write, delete, modify), and all operations work successfully.

8. IDENTIFY AND KILL HIGH CPU PROCESS

● SITUATION
The Linux server is running slow and you suspect that a process is consuming high CPU usage.

● METHOD 1: USING TOP COMMAND
top

Example Output
PID USER CPU COMMAND
2541 root 98 python

The process using highest CPU appears at the top.

Exit top
q

● METHOD 2: USING HTOP COMMAND
htop

Install htop (if not available)
sudo apt install htop
OR
sudo yum install htop

● METHOD 3: USING PS COMMAND
ps -eo pid,user,%cpu,%mem,command --sort=-%cpu | head

Example Output
PID USER %CPU COMMAND
2541 root 97.8 python
1752 root 30 java
1211 mysql 8 mysqld

● GRACEFUL TERMINATION (NORMAL KILL)
kill 2541

This sends SIGTERM signal
→ Process gets time to close safely and save data

● FORCE KILL
kill -9 2541

This sends SIGKILL signal
→ Process stops immediately without saving anything

● VERIFY PROCESS IS STOPPED
ps -ef | grep python
OR
top

● RESULT
The high CPU consuming process is identified using tools like top, htop, and ps. The process is then terminated using kill or kill -9, and system performance is restored.
